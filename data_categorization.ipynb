```python
import pandas as pd
```

```python
# df = pd.read_excel('data/full_marked.xlsx', sheet_name='Data')
# df = df[['name', 'target', 'store']].where(df['target'] != -1).dropna()
# df.columns

df = pd.read_csv('data/full_marked.csv')
df = df[['name', 'target', 'store']].where(df['target'] != -1).dropna()
df.columns
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

target_counts = df['target'].value_counts().sort_values(ascending=False)

# 2. Построение графика
plt.figure(figsize=(12, 6))
sns.set_style("whitegrid")

# Столбчатая диаграмма
ax = sns.barplot(x=target_counts.index, y=target_counts.values, palette="viridis")

# Подписи
plt.title('Распределение товаров по категориям (target)', fontsize=16, pad=20)
plt.xlabel('Категория (target)', fontsize=12)
plt.ylabel('Количество товаров', fontsize=12)
plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 10))
ax = sns.barplot(y=target_counts.values, x=target_counts.index, palette="mako")


plt.show()
```

```python
# Группировка: магазин + категория
store_target = df.groupby(['store', 'target']).size().unstack(fill_value=0)

# Тепловая карта
plt.figure(figsize=(10, 6))
sns.heatmap(store_target, annot=True, fmt='d', cmap='YlGnBu', cbar_kws={'label': 'Количество'})
plt.title('Распределение target по магазинам')
plt.xlabel('Категория (target)')
plt.ylabel('Магазин')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
