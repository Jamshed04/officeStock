```python
import pandas as pd
from tqdm import tqdm
import re
import nltk
from nltk.corpus import stopwords
import config.settings as settings
```

```python
root_path = settings.ROOT_DIR

goods_df = pd.read_excel(root_path / 'data/marked.xlsx', sheet_name='Data_new')
names_list = goods_df['name'].tolist()

nltk.download('stopwords', download_dir=root_path/'lib/nltk_data')
stop_words = set(stopwords.words('russian'))
for i, name in tqdm(enumerate(names_list), total=len(names_list)):
    name = names_list[i]
    text = name
    text = re.sub(r"«.*»", "", text)
    text = re.sub(r"([(][^)]+[)])|([\d]+\s?[в]\s?[\d+]|[№][\d]+)|(-\s) | (\*)", "", text)
    text = re.sub(r"([\d]+[.,]?[\d]+[%])|([\d]+[%])|(\d+[+])", "", text)
    text = re.sub(r"\d(?=[\.]|[\*]|[\,]).+", "", text)
    text = re.sub(r"\b\d+\s?(г|гр|гр\.|мл|л|шт|уп|пакет|%|штуки|штук|штука|банки)\b", "", text, flags=re.IGNORECASE)
    text = re.sub(r"[ ]{2,}", " ", text)
    words = [w for w in text.lower().split() if w not in stop_words]
    text = " ".join(words)
    names_list[i] = text.strip()
```
